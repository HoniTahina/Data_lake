
# TP4 — UC-07 : Architecture Azure

## 1. Infrastructure déployée

- Abonnement : Azure subscription 1 (free trial)
- Groupe de ressources : rg-retail-intelligence-tp4
- Compte de stockage : retailmia
- Région : France Central
- Type : Azure Storage avec ADLS Gen2
- Hierarchical Namespace (HNS) : activé
- Conteneur : retail-lake

## 2. Organisation du Data Lake

### Bronze
- Fichier : store.dat
- Origine : HDFS /retail-lake/raw/nifi/store.dat
- Contenu : données brutes des magasins

### Silver
- Jeu : uc06_store
- Format : Parquet
- Origine : transformations PySpark du TP3

### Gold
- Jeu : uc06_finance_magasins
- Format : Parquet
- Origine : agrégations PySpark du TP3

## 3. Modèle de sécurité

### Microsoft Entra ID
Service d'identification des utilisateurs et applications.

### Azure RBAC
Gestion des autorisations par rôles au niveau
des ressources Azure.

### ACL
Gestion des permissions sur les fichiers et répertoires
dans ADLS Gen2.

Les ACL d'accès concernent les objets existants.
Les ACL par défaut servent de modèle pour les
nouveaux éléments créés sous un répertoire.



## 4. Correspondance architecture locale / Azure

| Fonction | Architecture TP3 locale | Architecture Azure |
|---|---|---|
| Stockage | Hadoop HDFS | ADLS Gen2 |
| Ingestion | Apache NiFi | Azure Data Factory (proposé) |
| Transformation | PySpark local | Synapse Spark (proposé) |
| Données analytiques | Parquet sur HDFS | Parquet sur ADLS Gen2 |
| Analyse | JupyterLab et Spark SQL | Synapse SQL / Spark (proposé) |
| Visualisation | Matplotlib dans Jupyter | Power BI (facultatif) |
| Autorisations | Permissions Linux/HDFS | Entra ID, Azure RBAC et ACL |

## 5. Périmètre réellement réalisé

### Exécuté

## Vérification de la sécurité Azure

### Contrôle RBAC

L'examen de la section Access Control (IAM) du conteneur
retail-lake a permis d'identifier :

- Un rôle Owner hérité de l'abonnement.
- Un rôle User Access Administrator hérité d'un
  périmètre supérieur.

Ces rôles concernent la gestion des ressources
et des autorisations Azure.

###  Contrôle des ACL

Les permissions observées sur le répertoire racine
du conteneur sont :

| Principal | Permissions |
|---|---|
| Owner ($superuser) | rwx |
| Owning group ($superuser) | r-x |
| Other | --- |

Ces permissions correspondent à la configuration
visible dans le portail Azure.

Les ACL propres au répertoire gold restent
à vérifier séparément.
